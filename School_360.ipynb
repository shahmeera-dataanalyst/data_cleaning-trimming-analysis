FUNCTIONS

```python
def calculate_score(responses):
    total = 0

    for value in responses.values():   # LOOP
        total += value

    return total
```

```python
def classify_risk(score):

    if score <= 11:
        return "Low Risk"  # Changed from "Normal or Low Risk" for consistency

    elif score <= 15:
        return "Medium Risk"
    else:
        return "High Risk, Further evaluation needed"
```

```python
questions = [
    "Often complains of headaches, stomach-aches or sickness",
    "Many worries, often seems worried",
    "Often unhappy, down-hearted or tearful",
    "Nervous or clingy in new situations, easily loses confidence",
    "Many fears or easily scared",
    "Often has temper tantrums or hot tempers",
    "Often fights with other children or bullies them",
    "Often lies or cheats",
    "Steals from home, school or elsewhere",
    "Restless or over-active",
    "Constantly fidgeting or squirming",
    "Easily distracted or concentration wanders",
    "Rather solitary, tends to play alone",
    "Picked on or bullied by other children",
    "Gets on better with adults than with other children",
    "Generally obedient, usually does what adults request",
    "Thinks things out before acting",
    "Sees tasks through to the end, good attention span",
    "Has at least one good friend",
    "Generally liked by other children"
]
```

```python
reverse_questions = [
    "Generally obedient, usually does what adults request",
    "Thinks things out before acting",
    "Sees tasks through to the end, good attention span",
    "Has at least one good friend",
    "Generally liked by other children"
]
```

```python
print(len(reverse_questions))
```

```python
options = {
    "Not true": 0,
    "Somewhat true": 1,
    "Certainly true": 2
}
```

SCORING LOGIC

```python
scores = {}
st.subheader("SDQ_T_Questionnaire")

for q in questions:
    ans = st.radio(q, list(options.keys()), key=q)
    raw = options[ans]

    if q in reverse_questions:
        score = 2 - raw
    else:
        score = raw

    scores[q] = score

total_score = sum(scores.values())
```

RESULTS AND RECOMMENDATIONS

```python
if st.button("Calculate Result"):
    st.subheader("SDQ_T_Results")
    score = calculate_score(scores)

    category = classify_risk(score)

    st.success(f"Total Score: {score}")
    st.info(f"Risk Category: {category}")


    st.subheader("Mental_Health_Screening_Recommendation")

    if category == "Low Risk":
        recommendation = "Continue healthy habits, scheduled longitudinal monitoring recommended."

    elif category == "Medium Risk":
        recommendation = "Monitor emotional wellbeing and seek support if needed."

    else:
        recommendation = "Professional counselling referral is recommended."

    st.write(recommendation)

    st.session_state["sdq_total"] = score
    st.session_state["sdq_category"] = category
    st.session_state["sdq_recommendation"] = recommendation
    st.session_state["sdq_completed"] = True
```

```python
import pandas as pd
boys_ref=pd.read_csv("boys_ref.csv")
girls_ref=pd.read_csv("girls_ref.csv")
```

BMI_FUNCTION

```python
st.subheader("BMI Assessment")

age = st.number_input(
    "Age (Years)",
    min_value=5,
    max_value=19,
    step=1
)

gender = st.selectbox(
    "Gender",
    ["Male", "Female"]
)

height_cm = st.number_input(
    "Height (cm)",
    min_value=50.0,
    max_value=250.0,
    step=0.1
)

weight_kg = st.number_input(
    "Weight (kg)",
    min_value=1.0,
    max_value=200.0,
    step=0.1
)
```

BMI_CATEGORY

```python
def bmi_category(age, gender, bmi):

    age_months = age * 12

    if gender == "Male":
        ref = boys_ref
    else:
        ref = girls_ref

    row = ref.iloc[
        (ref["Month"] - age_months).abs().argsort()[:1]
    ]

    sd2neg = float(row["SD2neg"].iloc[0])
    sd1 = float(row["SD1"].iloc[0])
    sd2 = float(row["SD2"].iloc[0])

    if bmi < sd2neg:
        return "Underweight"

    elif bmi <= sd1:
        return "Normal"

    elif bmi <= sd2:
        return "Overweight"

    else:
        return "Obese"
```

BMI_RESULT_AND_RECOMMENDATION

```python
def calculate_bmi(weight_kg, height_cm):
    return weight_kg / ((height_cm / 100) ** 2)


if st.button("Calculate BMI"):

    bmi = calculate_bmi(
        weight_kg,
        height_cm
    )

    category = bmi_category(
        age,
        gender,
        bmi
    )

    st.success(f"BMI = {bmi:.2f}")

    st.info(f"Category = {category}")

    st.subheader("BMI Recommendation")

    if category == "Normal":
        recommendation = (
            "Healthy growth pattern observed. "
            "Continue balanced nutrition, regular physical activity, "
            "and routine health monitoring."
        )

    elif category == "Underweight":
        recommendation = (
            "Nutritional assessment recommended. "
            "Review dietary intake and monitor growth pattern. "
            "Consider referral for medical evaluation if concerns persist."
        )

    elif category == "Overweight":
        recommendation = (
            "Lifestyle and dietary counselling recommended. "
            "Encourage healthy eating habits, increased physical activity, "
            "and periodic weight monitoring."
        )

    else:
        recommendation = (
            "Medical evaluation and weight management counselling recommended. "
            "Comprehensive assessment and follow-up are advised."
        )

    st.write(recommendation)

    # Save results for summary report
    st.session_state["bmi_value"] = round(bmi, 1)
    st.session_state["bmi_result"] = category
    st.session_state["bmi_recommendation"] = recommendation
    st.session_state["bmi_completed"] = True
```

BP_REFERENCE_DICTIONARIES

```python
bp_boys = {

    5: {"SBP": 106, "DBP": 63},
    6: {"SBP": 107, "DBP": 68},
    7: {"SBP": 109, "DBP": 70},
    8: {"SBP": 110, "DBP": 71},
    9: {"SBP": 110, "DBP": 73},
    10: {"SBP": 112, "DBP": 74},
    11: {"SBP": 114, "DBP": 75},
    12: {"SBP": 117, "DBP": 75}

}

bp_girls = {

    5: {"SBP": 107, "DBP": 64},
    6: {"SBP": 108, "DBP": 68},
    7: {"SBP": 109, "DBP": 70},
    8: {"SBP": 110, "DBP": 71},
    9: {"SBP": 111, "DBP": 72},
    10: {"SBP": 112, "DBP": 73},
    11: {"SBP": 114, "DBP": 74},
    12: {"SBP": 116, "DBP": 75}

}
```

BP_FUNCTION

```python
def bp_screening(age, gender, sbp, dbp):

    if age >= 13:

        if sbp < 120 and dbp < 80:
            return "Normal"

        return "Abnormal"

    if gender == "Male":

        cutoff_sbp = bp_boys[age]["SBP"]
        cutoff_dbp = bp_boys[age]["DBP"]

    else:

        cutoff_sbp = bp_girls[age]["SBP"]
        cutoff_dbp = bp_girls[age]["DBP"]

    if sbp >= cutoff_sbp or dbp >= cutoff_dbp:

        return "Abnormal"

    return "Normal"
```

BLOOD_PRESSURE_RESULT_AND_RECOMMENDATION

```python
if st.button("Assess Blood Pressure"):

    bp_result = bp_screening(
        age,
        gender,
        sbp,
        dbp
    )

    st.subheader("Blood Pressure Screening Result")

    st.write(f"Systolic BP: {sbp} mmHg")

    st.write(f"Diastolic BP: {dbp} mmHg")

    st.write(f"Status: {bp_result}")




    st.subheader("BP_Recommendation")

    if bp_result == "Normal":

        recommendation_text = (
            "Blood pressure is within age-specific screening limits. Continue routine monitoring during school health assessments."
        )
        st.success(recommendation_text)

    else:

        recommendation_text = (
            "Blood pressure exceeds age-specific screening thresholds. Repeat BP measurement and seek pediatric assessment if elevated on repeat measurement."
        )
        st.warning(recommendation_text)
    st.session_state["bp_result"] = bp_result
    st.session_state["bp_recommendation"] = recommendation_text
    st.session_state["bp_completed"] = True
```

```python
st.header("Overall Health Status")

def traffic_light(result):

    if result is None:
        return "⚪"

    result = str(result).lower()

    green_words = [
        "Normal",
        "Healthy",
        "Pass",
        "Low Risk",
        "Normal Vision",
        "Adequate Color Vision",
        "Healthy Hair and Scalp",
        "Good Oral Health",
        "Healthy Skin",
        "Healthy Nails"
    ]

    yellow_words = [
        "Borderline",
        "Medium Risk",
        "Hair Hygiene Concern",
        "Nail Hygiene Concern",
        "Mild",
        "Overweight",
        "Monitor",
        "Fair",

      ]

    red_words = [
        "Abnormal",
        "High Risk",
        "Underweight",
        "Obese",
        "Under Weight",
        "Vision Referal Recommended",
        "Further Assessment Recommended",
        "Referal Recommended",
        "Skin Health Concern",
        "Nail Health Concern",
        'Poor',
        "Oral Health Concern"
        ]

    if any(word in result for word in green_words):
        return "🟢"

    elif any(word in result for word in yellow_words):
        return "🟡"

    elif any(word in result for word in red_words):
        return "🔴"

    else:
        return "⚪"


dashboard = {

    "BMI":
        traffic_light(
            st.session_state.get("bmi_result")
        ),

    "Blood Pressure":
        traffic_light(
            st.session_state.get("bp_result")
        ),

    "Vision":
        traffic_light(
            st.session_state.get("vision_result")
        ),

    "Color Vision":
        traffic_light(
            st.session_state.get("color_vision_result")
        ),

    "Hair":
        traffic_light(
            st.session_state.get("hair_result")
        ),

    "Oral":
        traffic_light(
            st.session_state.get("oral_result")
        ),

    "Skin":
        traffic_light(
            st.session_state.get("skin_result")
        ),

    "Nails":
        traffic_light(
            st.session_state.get("nail_result")
        ),

    "SDQ":
        traffic_light(
            st.session_state.get("sdq_category")
        )
}

for module, status in dashboard.items():

    st.write(
        f"{status}  {module}"
    )
```

```python

# Referral Tracking


st.subheader("Referral Generation and Tracking")

referrals = []

# Mental Health

if st.session_state.get("sdq_completed"):

    if st.session_state.get("sdq_category") in [


        "High Risk"

    ]:

        referrals.append(
            "Professional counselling referral is recommended."
        )

# BMI

if st.session_state.get("bmi_completed"):

    if st.session_state.get("bmi_result") in [

        "UNDERWEIGHT",
        "OBESE"

    ]:

        referrals.append(
             "Nutritional assessment recommended. "
            "Review dietary intake and monitor growth pattern. "
            "Consider referral for medical evaluation if concerns persist."
            "Medical evaluation and weight management counselling recommended. "
            "Comprehensive assessment and follow-up are advised."
        )

# Blood Pressure

if st.session_state.get("bp_completed"):

    if st.session_state.get("bp_result") != "Normal":

        referrals.append(
            "Blood pressure exceeds age-specific screening thresholds. Repeat BP measurement and seek pediatric assessment if elevated on repeat measurement."
        )

# Visual Acuity

if st.session_state.get("vision_completed"):

    if "referral" in str(
        st.session_state.get("vision_result")
    ).lower():

        referrals.append(
            """
Visual acuity is outside normal screening limits.

Referral to an ophthalmologist or optometrist is recommended for further evaluation.
"""
        )

# Colour Vision

if st.session_state.get("color_vision_completed"):

    if "deficiency" in str(
        st.session_state.get("color_vision_result")
    ).lower():

        referrals.append(
            """"
Possible color vision deficiency detected.

Referral for formal ophthalmologic assessment is recommended.
"""
        )

# Oral Health

if st.session_state.get("oral_completed"):

    if st.session_state.get("oral_result") != "Normal":

        referrals.append(
            """
Oral health concern detected.

Dental assessment is recommended.
"""
        )

# Skin

if st.session_state.get("skin_completed"):

    if st.session_state.get("skin_result") != "Normal":

        referrals.append(
            """
Unclassified or significant abnormal skin finding.

Further clinical assessment is recommended.
"""
        )

# Hair

if st.session_state.get("hair_completed"):

    if st.session_state.get("hair_result") != "Normal":

        referrals.append(
             """
Nits/lice observed.

Regular wet combing. Over-the-counter anti-lice shampoo advised.

Abnormal finding detected.

Further clinical assessment is recommended.
"""
        )

# Nails

if st.session_state.get("nail_completed"):

    if st.session_state.get("nail_result") != "Normal":

        referrals.append(
            """
Abnormal nail findings detected.

Further clinical assessment is recommended.
Unclassified or significant nail abnormality detected.

Further clinical assessment is recommended.
"""

     )

if referrals:

    for ref in referrals:

        st.warning(ref)

else:

    st.success(
        "No additional referrals suggested based on current screening findings."
    )
```
